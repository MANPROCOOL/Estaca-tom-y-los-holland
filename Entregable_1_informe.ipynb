Propuesta de pregunta de investigación formal:

Pregunta: ¿Existe asociación entre el tipo de alojamiento (room_type) ofrecido y el grupo de barrio (neighbourhood_group) en que se ubica, en los listados de Airbnb de Nueva York?

H₀: El tipo de alojamiento es independiente del barrio (no hay asociación).
H₁: El tipo de alojamiento está asociado al barrio (la distribución de tipos varía significativamente entre barrios).

Test estadístico: Chi-cuadrado de independencia → resultado preliminar: χ² = 2708.85, p < 0.001, se rechaza H₀.
