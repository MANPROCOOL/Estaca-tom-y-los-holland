## 1. Introducción y Motivación del Dataset

El mercado de alojamiento temporal ha transformado de manera profunda la dinámica urbana, turística e inmobiliaria de las principales metrópolis globales. En particular, la ciudad de Nueva York representa uno de los centros turísticos y financieros más dinámicos del mundo, donde la convivencia entre residentes permanentes y la oferta de arriendos a corto plazo genera constantes debates sobre acceso a la vivienda, regulación municipal y precios.

Para este laboratorio, seleccionamos el dataset público **Airbnb Open Data (NYC)**, disponible en la plataforma Kaggle. El conjunto de datos original cuenta con **102.599 registros y 26 variables**, abarcando dimensiones espaciales (distrito, barrio, coordenadas), económicas (precio por noche, tarifa de servicio), operacionales (noches mínimas, disponibilidad anual, política de cancelación) y de satisfacción de usuario (calificación del listado y cantidad de reseñas).

A diferencia de datasets sintéticos o excesivamente curados, este conjunto de datos presenta desafíos de calidad genuinos: registros duplicados, valores tipográficos inconsistentes, variables numéricas almacenadas con formatos de texto y símbolos monetarios, y valores fuera de rango físico. Esto ofrece un terreno idóneo para aplicar un flujo riguroso de Análisis Exploratorio de Datos (EDA) y validación de hipótesis estadísticas.

## 3. Justificación Metodológica de la Limpieza de Datos

Para asegurar la máxima validez y evitar distorsiones en las conclusiones estadísticas, el equipo adoptó un enfoque de limpieza que preserva la integridad del dataset completo (**102.058 observaciones netas** tras deduplicación), evitando la eliminación masiva de filas (`dropna`) y absteniéndose de imputaciones artificiales (como rellenar con medianas o medias) que reducen la varianza natural o alteran la distribución empírica.

Las decisiones de depuración implementadas en el código base son:

1. **Eliminación de duplicados exactos:** Se identificaron y eliminaron **541 filas duplicadas exactas** (mismo identificador y atributos idénticos), pasando de 102.599 a 102.058 registros.
2. **Corrección de errores tipográficos:** En la variable geográfica fundamental `neighbourhood_group`, se unificaron errores de digitación de texto: `"manhatan"` fue reemplazado por `"Manhattan"` y `"brookln"` por `"Brooklyn"`, consolidando exactamente los 5 distritos oficiales de la ciudad.
3. **Estandarización de variables monetarias:** Las columnas `price` y `service_fee` venían originalmente almacenadas como cadenas de texto con símbolos monetarios y comas (ej. `"$966 "`). Se despojaron de estos caracteres y se convirtieron a tipo numérico de coma flotante (`float`), manteniendo explícitamente los valores `NaN` para no imputar valores inexistentes.
4. **Normalización de nombres de columnas:** Se renombraron todas las cabeceras a minúsculas y formato `snake_case` (reemplazando espacios por guiones bajos), previniendo errores de referencia en el código.
5. **Eliminación de columnas no informativas:** Se eliminaron 4 columnas que no aportaban valor analítico:
   * `license`: contenía 102.597 valores nulos sobre 102.599 filas (~100% vacía).
   * `house_rules`: más del 51% de nulos y formato de texto libre desestructurado.
   * `country` y `country_code`: invariantes (~100% "United States" y "US"), carentes de variabilidad clasificatoria.
6. **Corrección de valores físicamente imposibles a `NaN`:**
   * `availability_365`: Por definición física, la disponibilidad anual de un alojamiento no puede ser inferior a 0 ni superior a 365 días. Se detectaron **431 registros negativos y 2.754 registros con valores superiores a 365 días (total: 3.185 registros inválidos)**. En lugar de eliminar estas filas o "adivinar" el valor, se convirtieron rigurosamente a `NaN`, preservando el resto de las variables del anuncio.
   * `minimum_nights`: Se detectaron **13 registros con valores negativos**, lo cual carece de sentido lógico. Fueron igualmente corregidos a `NaN`. Los valores muy altos (ej. arriendos $>365$ días) no se alteraron aquí, ya que pueden corresponder a contratos de arriendo residencial de largo plazo válidos en el mercado, dejándose para el análisis cuantitativo de outliers por IQR.

## 4. Estadística Descriptiva y Detección de Outliers (Criterio IQR)

Para cumplir con el criterio cuantitativo exigido en la identificación de valores atípicos, se aplicó la regla del **Rango Intercuartílico (IQR)** sobre las variables numéricas clave, definiendo los límites inferior y superior como:

$$\text{Límite Inferior} = Q_1 - 1.5 \cdot \text{IQR}, \quad \text{Límite Superior} = Q_3 + 1.5 \cdot \text{IQR}, \quad \text{donde } \text{IQR} = Q_3 - Q_1$$

Los resultados cuantitativos obtenidos sobre el dataset limpio son:

| Variable | $Q_1$ | $Q_3$ | $\text{IQR}$ | Límite Inferior | Límite Superior | Outliers Detectados | Comportamiento Estadístico |
| :--- | :---: | :---: | :---: | :---: | :---: | :---: | :--- |
| **price** | \$340.0 | \$913.0 | \$573.0 | -\$519.5 | \$1.772.5 | **0** | Distribución acotada de manera casi uniforme entre \$50 y \$1.200. |
| **service_fee** | \$68.0 | \$183.0 | \$115.0 | -\$104.5 | \$355.5 | **0** | Proporcional a `price` (~20%), perfectamente acotada. |
| **availability_365** | 2.0 d | 254.0 d | 252.0 d | -376.0 d | 632.0 d | **0** | Rango empírico $[0, 365]$ tras depuración de valores imposibles a NaN. |
| **minimum_nights** | 2.0 n | 5.0 n | 3.0 n | -2.5 n | 9.5 n | **18.240** | Fuerte asimetría positiva con cola pesada (máximo: 5.645 noches). |
| **number_of_reviews** | 1.0 | 31.0 | 30.0 | -44.0 | 76.0 | **11.120** | Distribución asimétrica típica de actividad en plataformas web. |

**Interpretación de Outliers:**
* En `price`, `service_fee` y `availability_365`, la ausencia de outliers IQR confirma que sus distribuciones son acotadas y carecen de valores extremos aislados.
* En `minimum_nights` y `number_of_reviews`, la detección de un número significativo de outliers refleja la naturaleza asimétrica y de ley de potencias inherente a estos procesos: la gran mayoría de los alojamientos exigen estancias cortas (1 a 4 noches) y reciben pocas reseñas, mientras que un grupo menor opera como alquiler corporativo/mensual o concentra un tráfico masivo de huéspedes.

## 5. Análisis Visual y Hallazgos Clave ("Observar con Lupa")

Al realizar la exploración gráfica tanto univariada como bivariada mediante Matplotlib y Seaborn, emergen varios hallazgos que merecen una inspección crítica y reflexiva:

### 5.1 Concentración Geográfica y Oferta Inmobiliaria
* **Dominio absoluto de Manhattan y Brooklyn:** Juntos concentran el **83.3%** del total de alojamientos listados en la ciudad (Manhattan con 43.792 listados y Brooklyn con 41.838 listados). Por el contrario, distritos como Staten Island representan menos del 1% (955 listados).
* **Composición del tipo de habitación:** El mercado se encuentra dominado por dos modalidades principales: departamentos o casas completas (*Entire home/apt*, 53.088 listados) y habitaciones privadas (*Private room*, 46.556 listados). Las habitaciones compartidas (*Shared room*, 2.226) y las habitaciones de hotel (*Hotel room*, 115) son minoritarias, estando estas últimas prácticamente confinadas a Manhattan (100 de 115).

### 5.2 "Observar con Lupa": El Comportamiento Inusual de Precios y Calificaciones
Al cruzar las variables económicas y de satisfacción contra los distritos y características de la propiedad, se observaron patrones contra-intuitivos que revelan aspectos clave sobre la naturaleza del dataset:

1. **Precios independientes del barrio y del tipo de habitación:** Los gráficos de cajas (*boxplots*) de `price` desglosados por `neighbourhood_group` revelan medianas y rangos intercuartílicos virtualmente indistinguibles entre distritos tan dispares económicamente como Manhattan (mediana \$628) y el Bronx (mediana \$623). Lo mismo ocurre al comparar un departamento entero frente a una habitación privada. En un mercado inmobiliario real, las propiedades completas en Manhattan exhiben precios sustancialmente superiores a las habitaciones en la periferia. Este comportamiento casi plano indica una fuerte posibilidad de que los precios en este dataset no correspondan a transacciones de mercado orgánicas o hayan sido anonimizados/sintetizados uniformemente.
2. **Homogeneidad en las calificaciones (`review_rate_number`):** Las calificaciones otorgadas por los huéspedes se distribuyen en una escala discreta de 1 a 5, pero al observar los promedios por distrito, todos fluctúan en un rango estrechísimo: Staten Island ($3.40$), Bronx ($3.33$), Queens ($3.33$), Manhattan ($3.28$) y Brooklyn ($3.26$). Además, pruebas bivariadas complementarias demostraron que `review_rate_number` no guarda relación con la política de cancelación (ANOVA $F = 0.0400, p = 0.9608$) ni con la reserva inmediata (Prueba $t = 1.0123, p = 0.3114$).

## 6. Pregunta de Investigación, Hipótesis y Validación Estadística

A partir de las observaciones anteriores, consolidamos la investigación en torno a la relación entre la ubicación geográfica y la satisfacción declarada por los huéspedes.

### 6.1 Pregunta de Investigación Formal
> **¿Existe una asociación estadísticamente significativa entre el distrito geográfico (`neighbourhood_group`) y la calificación otorgada por los huéspedes (`review_rate_number`) en los alojamientos de Airbnb en la ciudad de Nueva York?**

### 6.2 Formulación de Hipótesis
* **Hipótesis Nula ($H_0$):** La calificación promedio del alojamiento es independiente del distrito donde se ubica; no existen diferencias en la distribución ni en la media de `review_rate_number` entre los 5 distritos de Nueva York:
  $$\mu_{\text{Bronx}} = \mu_{\text{Brooklyn}} = \mu_{\text{Manhattan}} = \mu_{\text{Queens}} = \mu_{\text{Staten Island}}$$
* **Hipótesis Alternativa ($H_1$):** La calificación promedio del alojamiento no es independiente del distrito; existe una diferencia estadísticamente significativa en la calificación media en al menos un distrito respecto a los demás:
  $$\exists \, i, j \quad \text{tal que} \quad \mu_i \neq \mu_j$$

Nivel de significancia adoptado: $\alpha = 0.05$.

---

### 6.3 Test Principal: Análisis de Varianza Unidireccional (ANOVA)

Para responder a la pregunta de investigación principal, se realizó un **ANOVA de una vía** comparando las medias de `review_rate_number` a través de los 5 distritos independientes (`neighbourhood_group`):

* **Estadístico $F$ calculado:** $F = 11.3048$
* **$p$-valor:** $p = 3.596 \times 10^{-9} \approx 0.0000$
* **Descriptivos por grupo:**
  * Staten Island: $N = 943, \, \bar{x} = 3.4040, \, s = 1.2556$
  * Bronx: $N = 2.677, \, \bar{x} = 3.3313, \, s = 1.2126$
  * Queens: $N = 13.159, \, \bar{x} = 3.3304, \, s = 1.2486$
  * Manhattan: $N = 43.418, \, \bar{x} = 3.2765, \, s = 1.2907$
  * Brooklyn: $N = 41.515, \, \bar{x} = 3.2585, \, s = 1.2950$

**Decisión estadística:** Dado que $p < \alpha$ ($p \approx 0.0000 < 0.05$), se **rechaza la hipótesis nula ($H_0$)** a favor de la hipótesis alternativa ($H_1$). Desde una perspectiva estadística formal, existe una diferencia en las calificaciones medias entre distritos.

---

### 6.4 Medida Complementaria de Tamaño de Efecto: Chi-cuadrado y V de Cramér

Reportar únicamente el $p$-valor en muestras masivas es metodológicamente insuficiente, ya que un tamaño de muestra $N > 100.000$ otorga una potencia estadística tan elevada que vuelve "significativa" cualquier desviación trivial. Para evaluar la **relevancia práctica** de esta relación, tratamos `review_rate_number` como una variable discreta/ordinal en escala 1 a 5 y evaluamos la tabla de contingencia de $5 \times 5$ ($N = 101.712$ pares completos):

* **Prueba Chi-cuadrado de Independencia:**
  * $\chi^2 = 162.5295$
  * Grados de libertad: $df = (5 - 1) \times (5 - 1) = 16$
  * $p$-valor: $p = 2.586 \times 10^{-26} \approx 0.0000$
  * Mínimo valor esperado en celdas: $85.12$ (cumple holgadamente el criterio de validez de Cochran $\ge 5$).
* **Cálculo del Coeficiente V de Cramér:**
  $$V = \sqrt{\frac{\chi^2}{N \cdot \min(r-1, c-1)}} = \sqrt{\frac{162.5295}{101.712 \cdot 4}} = \sqrt{\frac{162.5295}{406.848}} \approx \mathbf{0.0200}$$

*(Nota metodológica: En exploraciones previas sobre submuestras con imputación y filtrado agresivo se obtuvo $V = 0.0523$; en ambos casos, el coeficiente cae de lleno en la misma categoría sustantiva).*

**Conclusión Estadística e Interpretación Rigurosa ("Mirar con Lupa"):**
La magnitud del efecto es **extremadamente débil / prácticamente nula ($V = 0.0200 \ll 0.10$)**. Esto demuestra con total rigor estadístico que, aunque la prueba ANOVA y la prueba Chi-cuadrado arrojen significancia estadística debido al enorme tamaño de la muestra ($N > 100.000$), **la relación carece de impacto práctico en el mundo real**: la diferencia entre la nota promedio del distrito mejor calificado (Staten Island, 3.40) y el peor calificado (Brooklyn, 3.26) es de apenas $0.14$ puntos en una escala de 1 a 5. Por lo tanto, el distrito donde se aloja el huésped **no es un factor determinante en la calificación que otorga**.

### 6.5 Variable Secundaria de Contexto: Tipo de Alojamiento (`room_type` vs `neighbourhood_group`)

Como variable secundaria de contraste estructural, evaluamos la relación entre el tipo de habitación (`room_type`) y el distrito (`neighbourhood_group`):

* **Prueba Chi-cuadrado:** $\chi^2 = 2708.85, \, p < 10^{-16}, \, df = 12$.
* **Interpretación:** A diferencia de las calificaciones, la estructura inmobiliaria sí exhibe una dependencia geográfica real y marcada: Manhattan concentra de forma casi exclusiva las habitaciones de hotel (*Hotel room*, 100 de 115) y presenta un claro predominio de departamentos completos (*Entire home/apt*, 54.4%), mientras que Brooklyn y Queens presentan una mayor proporción de habitaciones privadas (*Private room*, 42% y 54% respectivamente).

## 7. Limitaciones del Dataset

Para una correcta interpretación de los resultados y su alcance, es fundamental documentar las limitaciones que presenta este conjunto de datos:

1. **Representatividad Geográfica Asimétrica:** Existe una fuerte concentración en Manhattan y Brooklyn (~83% del total de registros), mientras que Staten Island representa menos del 1% (955 listados). Aunque el tamaño absoluto de Staten Island permite aplicar pruebas asintóticas válidas ($N > 30$), la desproporción muestral puede afectar la robustez en análisis desagregados por micro-vecindarios.
2. **Indicios de Generación o Modificación Sintética:** Variables clave como `price` y `review_rate_number` exhiben propiedades empíricas altamente anómalas (distribución de precios cuasi-uniforme entre \$50 y \$1.200 y ausencia de gradiente geográfico; calificaciones con distribuciones idénticas entre distritos y políticas de reserva). Esto sugiere fuertemente que ciertas variables del dataset público pudieron haber sido anonimizadas, modificadas con ruido artificial o sintetizadas en el origen, limitando su representatividad directa sobre el mercado real neoyorquino.
3. **Falta de Continuidad Temporal:** Aunque existe una columna de última reseña (`last_review`), el dataset corresponde a una fotografía estática y no a un panel longitudinal. No es posible modelar fluctuaciones estacionales de precios o disponibilidad (ej. demanda en verano vs invierno), ni capturar los efectos de los cambios regulatorios sobre alquileres de corta estadía aplicados por la ciudad de Nueva York en años recientes.
4. **Calidad de Entrada de Datos de Usuarios:** La presencia de errores manifiestos en la entrada de datos (como valores de disponibilidad superiores a 365 días o noches mínimas negativas) evidencia una ausencia de validaciones estrictas al momento del registro, lo que refuerza la necesidad crítica de fases de depuración y justificación explícita de valores nulos.
