# Entregable 1: Análisis Exploratorio de Datos - Airbnb Nueva York

## 1. Introducción y Motivación
Para este proyecto, hemos seleccionado el conjunto de datos "Airbnb Open Data" de Nueva York. La motivación principal detrás de esta elección es el impacto significativo que tienen las plataformas de alojamiento temporal en la economía y la dinámica urbana de las grandes ciudades. Nueva York, al ser uno de los destinos turísticos más importantes del mundo, ofrece un terreno de estudio amplio en contenido.

Nuestro objetivo es comprender cómo se estructura la oferta de alojamientos, analizando si existen patrones geográficos marcados que definan el tipo de hospedaje disponible según el distrito (borough).

### Limitaciones del Dataset
Antes de seguir con el análisis, es importante reconocer ciertas limitaciones en nuestros datos:
* **Representatividad:** El dataset solo refleja los alojamientos listados en Airbnb, lo cual no es representativo de todo el mercado inmobiliario o turístico de la ciudad.
* **Sesgo temporal:** Los datos representan una "foto" en un momento específico, por lo que no capturan la estacionalidad (ej. fluctuaciones de precios o disponibilidad entre verano e invierno).
* **Calidad de los datos ingresados:** Al ser datos generados por los usuarios, pueden existir sesgos descriptivos, precios atípicos o errores de digitación.

## Motivación del dataset

El dataset **Airbnb Open Data (NYC)** fue obtenido de Kaggle
(kaggle.com/datasets/arianazmoudeh/airbnbopendata), una plataforma pública de
datasets recopilados y compartidos por la comunidad. Contiene registros de
listados de alojamiento publicados en la plataforma Airbnb en la ciudad de
Nueva York.

Se eligió este dataset porque el mercado de alojamiento de corto plazo es un
fenómeno económico y social relevante: permite estudiar cómo factores
geográficos (barrio), estructurales (tipo de alojamiento) y de percepción de
los huéspedes (calificaciones) se relacionan entre sí. Además, al tratarse de
datos reales sin procesar en exceso, ofrece la oportunidad de aplicar un
proceso de limpieza genuino (duplicados, errores de tipeo, formatos de texto
en variables numéricas, valores fuera de rango físico), a diferencia de
datasets ya excesivamente curados.

## Limitaciones del dataset

- **Representatividad:** los listados están fuertemente concentrados en
  Manhattan y Brooklyn (~85% del total), mientras que Staten Island representa
  menos del 1%. Esto puede afectar la robustez de comparaciones entre grupos
  con tamaños muestrales muy dispares.
- **Posibles sesgos en la recolección:** se detectó que variables como
  `price` y `review_rate_number` presentan patrones estadísticamente
  inusuales (distribución de precios prácticamente uniforme entre $50 y
  $1.200, y conteos casi idénticos entre categorías de rating 2 a 5), lo que
  sugiere que podrían no reflejar comportamiento real de mercado, sino haber
  sido generadas o modificadas artificialmente en algún punto de la
  recopilación del dataset.
- **Período/contexto de recolección:** el dataset no especifica con precisión
  el rango temporal exacto en que fueron recolectados los listados; la
  columna `last_review` sugiere reseñas entre 2018 y 2022 aproximadamente,
  pero no hay certeza sobre cuándo se creó cada listado ni si refleja
  condiciones de mercado actuales.
